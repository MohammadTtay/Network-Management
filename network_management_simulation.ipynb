# شبیه‌سازی اقتصادی مدیریت شبکه با Python

این نوت‌بوک اسکریپت اصلی پروژه را اجرا می‌کند. ابتدا `requirements.txt` را نصب کنید.

برای دیدن تمام منطق پیاده‌سازی، فایل `network_management_simulation.py` در همین پوشه قرار دارد. داده‌های تصادفی با `seed=42` قابل‌بازتولید هستند.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Image
from network_management_simulation import run_simulation

OUTPUT_DIR = Path('output')
result = run_simulation(seed=42, output_dir=OUTPUT_DIR)
display(result.metrics)
print((OUTPUT_DIR / 'network_management_report.txt').read_text(encoding='utf-8'))
```

## نمودارهای اصلی پروژه
